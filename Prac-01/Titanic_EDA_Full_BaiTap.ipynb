# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

````python


cols = ['age', 'fare', 'sibsp', 'parch', 'family_size']

central = pd.DataFrame({
    'Mean': df[cols].mean(),
    'Median': df[cols].median(),
    'Mode': [df[col].mode().iloc[0] for col in cols]
})

central
````

## Group 2 — Dispersion

````python

dispersion = pd.DataFrame({
    'Min': df[cols].min(),
    'Max': df[cols].max(),
    'Range': df[cols].max() - df[cols].min(),
    'Variance': df[cols].var(),
    'Std': df[cols].std(),
    'IQR': df[cols].quantile(0.75) - df[cols].quantile(0.25)
})

dispersion
````

## Group 3 — Location and Shape

````python
# TODO

shape_stats = pd.DataFrame({
    'Q1': df[cols].quantile(0.25),
    'Median_Q2': df[cols].quantile(0.50),
    'Q3': df[cols].quantile(0.75),
    'Skewness': df[cols].skew(),
    'Kurtosis': df[cols].kurt()
})

shape_stats
````

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

````python
# TODO
survival_by_class = df.groupby('class')['survived'].mean().sort_values(ascending=False)

print(survival_by_class)

highest = survival_by_class.iloc[0]
lowest = survival_by_class.iloc[-1]

print('Hạng vé sống sót cao nhất:', survival_by_class.index[0])
print('Hạng vé sống sót thấp nhất:', survival_by_class.index[-1])
print('Chênh lệch:', highest - lowest)
````

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

````python
# TODO
survival_by_sex = df.groupby('sex')['survived'].mean()
survival_by_class = df.groupby('class')['survived'].mean()

sex_diff = survival_by_sex.max() - survival_by_sex.min()
class_diff = survival_by_class.max() - survival_by_class.min()

print('Tỷ lệ sống theo giới tính:')
print(survival_by_sex)

print('\nTỷ lệ sống theo hạng vé:')
print(survival_by_class)

print('\nChênh lệch theo giới tính:', sex_diff)
print('Chênh lệch theo hạng vé:', class_diff)

if sex_diff > class_diff:
    print('\n=> Giới tính ảnh hưởng mạnh hơn.')
else:
    print('\n=> Hạng vé ảnh hưởng mạnh hơn.')
````

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

````python
# TODO
df['fare_group'] = pd.qcut(
    df['fare'],
    4,
    labels=['Low', 'Medium', 'High', 'Very High']
)

fare_survival = df.groupby('fare_group', observed=True)['survived'].mean()

print(fare_survival)
````

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

````python
# TODO
family_survival = df.groupby('family_size')['survived'].agg(
    ['count', 'mean']
)

print(family_survival)
````

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?
